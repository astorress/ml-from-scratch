# Predicción, Clasificación y Pronóstico

### ¿Cuál es la diferencia?

Cuando hablamos de que una computadora "aprende" de los datos, existen diferentes **tipos de tareas** que puede realizar. Tres de los conceptos más comunes — y más confundidos — son:

- **Predicción**
- **Clasificación**
- **Pronóstico**

Aunque los tres buscan "adivinar" algo desconocido a partir de datos conocidos, cada uno responde a una **pregunta diferente**.

Veamos una analogía cotidiana:

| Concepto | Pregunta que responde | Ejemplo cotidiano |
|---|---|---|
| **Predicción** | ¿Cuánto? ¿Qué valor? | "¿Cuánto costará esta casa?" |
| **Clasificación** | ¿Qué es? ¿A qué grupo pertenece? | "¿Este correo es spam o no spam?" |
| **Pronóstico** | ¿Qué pasará en el futuro? | "¿Lloverá la próxima semana?" |

Vamos a explorar cada uno con ejemplos claros, visualizaciones y datos reales.

---

## 1. Predicción

### ¿Qué es?

La **predicción** consiste en estimar un **valor numérico continuo** a partir de datos de entrada. El resultado es un **número**, no una categoría.

**Características:**
- La respuesta es un **número en un rango continuo** (puede ser 150,000, 150,001, 150,002...)
- Se basa en encontrar **relaciones** entre variables
- No necesariamente involucra el tiempo (a diferencia del pronóstico)

**Ejemplos:**

| Datos de entrada | Predicción (número) |
|---|---|
| Tamaño de casa, ubicación, habitaciones | Precio de la casa ($250,000) |
| Edad, peso, altura de un paciente | Presión arterial (120 mmHg) |
| Años de experiencia de un empleado | Salario ($45,000) |
| Ingredientes de una receta | Calorías (350 kcal) |

**La pregunta clave de la predicción siempre es: ¿CUÁNTO?**

```python
import numpy as np
import matplotlib.pyplot as plt

# =============================================
# EJEMPLO DE PREDICCIÓN:
# Predecir el precio de una casa según su tamaño
# =============================================

np.random.seed(42)

# Datos simulados: tamaño de casa (m²) vs precio (miles de $)
tamano = np.array([50, 60, 70, 80, 90, 100, 110, 120, 130, 140, 150, 160, 170, 180, 200])
precio = tamano * 1.8 + np.random.randn(len(tamano)) * 15 + 20

# Ajustar una línea (regresión lineal simple)
coef = np.polyfit(tamano, precio, 1)
linea = np.poly1d(coef)

# Casa nueva: ¿cuánto cuesta una de 135 m²?
casa_nueva = 135
precio_predicho = linea(casa_nueva)

fig, ax = plt.subplots(figsize=(10, 6))

# Datos conocidos
ax.scatter(tamano, precio, color='#3498db', s=100, edgecolors='k', zorder=5, label='Casas conocidas')

# Línea de tendencia
x_line = np.linspace(40, 210, 100)
ax.plot(x_line, linea(x_line), color='#e74c3c', linewidth=2, linestyle='--', label='Tendencia aprendida')

# Predicción
ax.scatter(casa_nueva, precio_predicho, color='#f39c12', s=300, marker='★', edgecolors='k',
           zorder=10, label=f'Predicción: ${precio_predicho:.0f}k')
ax.annotate(f'Casa de {casa_nueva} m²\n→ ${precio_predicho:.0f},000',
            xy=(casa_nueva, precio_predicho), xytext=(casa_nueva + 15, precio_predicho - 30),
            fontsize=11, fontweight='bold', color='#e67e22',
            arrowprops=dict(arrowstyle='->', color='#e67e22', lw=2),
            bbox=dict(boxstyle='round,pad=0.5', facecolor='#fdebd0', alpha=0.9))

ax.set_xlabel('Tamaño de la casa (m²)', fontsize=13)
ax.set_ylabel('Precio (miles de $)', fontsize=13)
ax.set_title('PREDICCIÓN: ¿Cuánto costará esta casa?', fontsize=15, fontweight='bold')
ax.legend(fontsize=11, loc='upper left')
ax.grid(True, alpha=0.3)

# Anotación de concepto
ax.text(160, 80, 'La respuesta es un\nNÚMERO CONTINUO\n(no una categoría)',
        fontsize=10, ha='center', style='italic',
        bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.8))

plt.tight_layout()
plt.show()

print(f"Pregunta: ¿Cuánto cuesta una casa de {casa_nueva} m²?")
print(f"Respuesta (predicción): ${precio_predicho:,.0f} (un número, no una categoría)")
```

Observa que la respuesta es **$263,000** — un número específico. No es "cara" o "barata", es un **valor numérico preciso**. Eso es predicción.

Otros ejemplos del mismo tipo:

- ¿Cuántos grados hará hoy? → **32.5 °C**
- ¿Cuántas calorías tiene este plato? → **487 kcal**
- ¿Qué nota sacará este alumno? → **7.8**

## 2. Clasificación

### ¿Qué es?

La **clasificación** consiste en asignar una **etiqueta o categoría** a un dato. El resultado no es un número, sino una **clase** a la que pertenece.

**Características:**
- La respuesta es una **categoría discreta** (grupo A, grupo B, grupo C...)
- Se basa en encontrar **fronteras** que separen los grupos
- No tiene relación con el tiempo

**Ejemplos:**

| Datos de entrada | Clasificación (categoría) |
|---|---|
| Texto de un correo electrónico | Spam / No spam |
| Imagen de un animal | Gato / Perro / Pájaro |
| Síntomas de un paciente | Sano / Enfermo |
| Transacción bancaria | Legítima / Fraudulenta |
| Foto de una fruta | Manzana / Naranja / Plátano |

**La pregunta clave de la clasificación siempre es: ¿QUÉ ES? ¿A QUÉ GRUPO PERTENECE?**

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap

# =============================================
# EJEMPLO DE CLASIFICACIÓN:
# Clasificar frutas por peso y tamaño
# =============================================

np.random.seed(42)

# Datos simulados de frutas
# Naranjas: pesadas y grandes
n = 20
naranjas_peso = np.random.normal(200, 30, n)
naranjas_diametro = np.random.normal(8, 1, n)

# Manzanas: medianas
manzanas_peso = np.random.normal(150, 25, n)
manzanas_diametro = np.random.normal(7, 0.8, n)

# Fresas: ligeras y pequeñas
fresas_peso = np.random.normal(30, 10, n)
fresas_diametro = np.random.normal(3, 0.7, n)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# --- Panel 1: Los datos ---
ax = axes[0]
ax.scatter(naranjas_peso, naranjas_diametro, c='#f39c12', s=120, edgecolors='k',
           label='Naranja 🍊', marker='o', zorder=5)
ax.scatter(manzanas_peso, manzanas_diametro, c='#e74c3c', s=120, edgecolors='k',
           label='Manzana 🍎', marker='s', zorder=5)
ax.scatter(fresas_peso, fresas_diametro, c='#c0392b', s=80, edgecolors='k',
           label='Fresa 🍓', marker='^', zorder=5)

# Fruta nueva
nueva_peso, nueva_diametro = 160, 7.2
ax.scatter(nueva_peso, nueva_diametro, c='lime', s=400, marker='★', edgecolors='k',
           zorder=10, label='¿Qué fruta es esta?')
ax.annotate('Fruta\ndesconocida', xy=(nueva_peso, nueva_diametro),
            xytext=(nueva_peso + 30, nueva_diametro - 1.5),
            fontsize=11, fontweight='bold', color='green',
            arrowprops=dict(arrowstyle='->', color='green', lw=2))

ax.set_xlabel('Peso (gramos)', fontsize=12)
ax.set_ylabel('Diámetro (cm)', fontsize=12)
ax.set_title('Los datos: frutas conocidas + una desconocida', fontsize=12, fontweight='bold')
ax.legend(fontsize=10)
ax.grid(True, alpha=0.3)

# --- Panel 2: La clasificación ---
ax = axes[1]

# Crear regiones de clasificación (simplificado)
all_peso = np.concatenate([naranjas_peso, manzanas_peso, fresas_peso])
all_diametro = np.concatenate([naranjas_diametro, manzanas_diametro, fresas_diametro])
all_labels = np.array([0]*n + [1]*n + [2]*n)

from sklearn.neighbors import KNeighborsClassifier
X_frutas = np.column_stack([all_peso, all_diametro])
clf = KNeighborsClassifier(n_neighbors=5)
clf.fit(X_frutas, all_labels)

h = 1
x_min, x_max = all_peso.min() - 20, all_peso.max() + 20
y_min, y_max = all_diametro.min() - 1, all_diametro.max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, h), np.arange(y_min, y_max, 0.1))
Z = clf.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)

cmap_light = ListedColormap(['#fdebd0', '#fadbd8', '#f5b7b1'])
ax.contourf(xx, yy, Z, alpha=0.4, cmap=cmap_light)

ax.scatter(naranjas_peso, naranjas_diametro, c='#f39c12', s=120, edgecolors='k', marker='o', zorder=5)
ax.scatter(manzanas_peso, manzanas_diametro, c='#e74c3c', s=120, edgecolors='k', marker='s', zorder=5)
ax.scatter(fresas_peso, fresas_diametro, c='#c0392b', s=80, edgecolors='k', marker='^', zorder=5)

# Clasificar la nueva fruta
pred = clf.predict([[nueva_peso, nueva_diametro]])[0]
nombres = ['Naranja 🍊', 'Manzana 🍎', 'Fresa 🍓']
colores = ['#f39c12', '#e74c3c', '#c0392b']

ax.scatter(nueva_peso, nueva_diametro, c=colores[pred], s=400, marker='★', edgecolors='k', zorder=10)
ax.annotate(f'¡Es una\n{nombres[pred]}!', xy=(nueva_peso, nueva_diametro),
            xytext=(nueva_peso + 30, nueva_diametro - 1.5),
            fontsize=12, fontweight='bold', color=colores[pred],
            arrowprops=dict(arrowstyle='->', color=colores[pred], lw=2),
            bbox=dict(boxstyle='round,pad=0.5', facecolor='white', alpha=0.9))

ax.set_xlabel('Peso (gramos)', fontsize=12)
ax.set_ylabel('Diámetro (cm)', fontsize=12)
ax.set_title('Clasificación: asignar a un GRUPO', fontsize=12, fontweight='bold')
ax.grid(True, alpha=0.3)

# Leyenda de regiones
ax.text(210, 4, 'Zona\nNaranja', fontsize=9, ha='center', color='#f39c12', fontweight='bold',
        bbox=dict(facecolor='white', alpha=0.7, boxstyle='round'))
ax.text(140, 9, 'Zona\nManzana', fontsize=9, ha='center', color='#e74c3c', fontweight='bold',
        bbox=dict(facecolor='white', alpha=0.7, boxstyle='round'))
ax.text(40, 3, 'Zona\nFresa', fontsize=9, ha='center', color='#c0392b', fontweight='bold',
        bbox=dict(facecolor='white', alpha=0.7, boxstyle='round'))

plt.suptitle('CLASIFICACIÓN: ¿Qué fruta es esta?', fontsize=15, fontweight='bold')
plt.tight_layout()
plt.show()

print(f"Pregunta: Tengo una fruta de {nueva_peso}g y {nueva_diametro}cm. ¿Qué fruta es?")
print(f"Respuesta (clasificación): ¡Es una {nombres[pred]}! (una categoría, no un número)")
```

Observa que la respuesta es **"Manzana"** — una categoría. No es "150 gramos" ni "7.2 cm", es una **etiqueta de grupo**. Eso es clasificación.

**Tipos de clasificación:**

| Tipo | Ejemplo |
|---|---|
| **Binaria** (2 clases) | Spam / No spam |
| **Multiclase** (3+ clases) | Gato / Perro / Pájaro |
| **Multietiqueta** (varias a la vez) | Una foto puede ser: [playa, atardecer, tropical] |

## 3. Pronóstico

### ¿Qué es?

El **pronóstico** (forecasting) consiste en estimar **valores futuros** basándose en **datos históricos ordenados en el tiempo**. El ingrediente clave que lo distingue es el **TIEMPO**.

**Características:**
- Los datos tienen un **orden temporal** (día 1, día 2, día 3...)
- Busca **patrones en el tiempo**: tendencias, estacionalidad, ciclos
- La pregunta siempre se refiere al **futuro**
- El pasado cercano suele ser más relevante que el pasado lejano

**Ejemplos:**

| Datos históricos | Pronóstico (futuro) |
|---|---|
| Ventas de los últimos 12 meses | Ventas del próximo mes |
| Temperatura de la última semana | Temperatura de mañana |
| Precio de una acción los últimos 30 días | Precio de mañana |
| Pacientes en urgencias por hora | Pacientes en las próximas 4 horas |

**La pregunta clave del pronóstico siempre es: ¿QUÉ PASARÁ EN EL FUTURO?**

```python
import numpy as np
import matplotlib.pyplot as plt

# =============================================
# EJEMPLO DE PRONÓSTICO:
# Predecir ventas futuras de una tienda
# =============================================

np.random.seed(42)

# Datos históricos: ventas mensuales de 2 años (24 meses)
meses = np.arange(1, 25)
# Tendencia ascendente + estacionalidad (más ventas en diciembre)
tendencia = meses * 2
estacionalidad = 15 * np.sin(2 * np.pi * meses / 12 - np.pi/2)  # pico en diciembre
ruido = np.random.randn(24) * 5
ventas = 80 + tendencia + estacionalidad + ruido

# Pronóstico: próximos 6 meses
meses_futuro = np.arange(25, 31)
tendencia_f = meses_futuro * 2
estacionalidad_f = 15 * np.sin(2 * np.pi * meses_futuro / 12 - np.pi/2)
ventas_pronostico = 80 + tendencia_f + estacionalidad_f

# Nombres de meses
nombres_meses = ['Ene','Feb','Mar','Abr','May','Jun','Jul','Ago','Sep','Oct','Nov','Dic'] * 3

fig, ax = plt.subplots(figsize=(14, 6))

# Datos históricos
ax.plot(meses, ventas, 'o-', color='#3498db', linewidth=2, markersize=6,
        label='Datos históricos (pasado)', zorder=5)

# Pronóstico
ax.plot(meses_futuro, ventas_pronostico, 's--', color='#e74c3c', linewidth=2.5, markersize=8,
        label='Pronóstico (futuro)', zorder=5)

# Banda de incertidumbre
incertidumbre = np.arange(1, 7) * 4  # crece con el tiempo
ax.fill_between(meses_futuro, ventas_pronostico - incertidumbre,
                ventas_pronostico + incertidumbre,
                color='#e74c3c', alpha=0.15, label='Incertidumbre')

# Línea divisoria HOY
ax.axvline(x=24.5, color='gray', linewidth=2, linestyle=':', zorder=3)
ax.text(23, 145, '← PASADO', fontsize=12, fontweight='bold', color='#3498db', ha='right')
ax.text(26, 145, 'FUTURO →', fontsize=12, fontweight='bold', color='#e74c3c')
ax.text(24.5, 150, 'HOY', fontsize=11, fontweight='bold', ha='center', color='gray',
        bbox=dict(facecolor='white', edgecolor='gray', boxstyle='round'))

# Etiquetas de meses
all_meses = list(meses) + list(meses_futuro)
ax.set_xticks(all_meses[::3])
ax.set_xticklabels([nombres_meses[m-1] + f"\n'{22 + m//12}" for m in all_meses[::3]], fontsize=8)

# Anotar estacionalidad
ax.annotate('Pico\ndiciembre', xy=(12, ventas[11]), xytext=(14, 65),
            fontsize=9, color='#8e44ad', fontweight='bold',
            arrowprops=dict(arrowstyle='->', color='#8e44ad'))

ax.set_xlabel('Mes', fontsize=12)
ax.set_ylabel('Ventas (unidades)', fontsize=12)
ax.set_title('PRONÓSTICO: ¿Cuánto venderemos los próximos meses?', fontsize=15, fontweight='bold')
ax.legend(fontsize=11, loc='upper left')
ax.grid(True, alpha=0.3)

# Caja de concepto
ax.text(6, 140, 'El pronóstico usa el\nPATRÓN TEMPORAL\n(tendencia + estacionalidad)\npara estimar el futuro',
        fontsize=9, ha='center', style='italic',
        bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.9))

plt.tight_layout()
plt.show()

print("Pregunta: ¿Cuánto venderemos en los próximos 6 meses?")
print("Respuesta (pronóstico): estimaciones MES a MES basadas en el patrón histórico")
print("\nNota: El pronóstico incluye INCERTIDUMBRE que crece a medida que miramos más lejos.")
```

Observa que el pronóstico tiene dos elementos clave que lo diferencian:

1. **Eje temporal**: los datos están ordenados en el tiempo (mes 1, mes 2, mes 3...)
2. **Incertidumbre creciente**: cuanto más lejos miramos, más incierta es la estimación (la banda roja se ensancha)

**Patrones que un pronóstico puede detectar:**

| Patrón | Descripción | Ejemplo |
|---|---|---|
| **Tendencia** | Dirección general (sube, baja, plano) | Las ventas suben 2% cada mes |
| **Estacionalidad** | Patrón que se repite periódicamente | Más helados en verano, más chocolate en invierno |
| **Ciclo** | Patrón irregular de largo plazo | Ciclos económicos de expansión y recesión |
| **Ruido** | Variación aleatoria impredecible | Fluctuaciones día a día sin patrón |

```python
import numpy as np
import matplotlib.pyplot as plt

# Visualizar los componentes de una serie temporal
np.random.seed(42)

t = np.arange(1, 49)  # 48 meses

tendencia = t * 1.5
estacionalidad = 20 * np.sin(2 * np.pi * t / 12)
ruido = np.random.randn(48) * 5
total = 50 + tendencia + estacionalidad + ruido

fig, axes = plt.subplots(4, 1, figsize=(14, 10), sharex=True)

# Total
axes[0].plot(t, total, 'o-', color='#2c3e50', markersize=3, linewidth=1.5)
axes[0].set_ylabel('Dato\nobservado', fontsize=10)
axes[0].set_title('Descomposición de una Serie Temporal', fontsize=14, fontweight='bold')
axes[0].grid(True, alpha=0.3)

# Tendencia
axes[1].plot(t, 50 + tendencia, color='#e74c3c', linewidth=2.5)
axes[1].set_ylabel('Tendencia', fontsize=10, color='#e74c3c')
axes[1].text(45, 60, '↗ Sube', fontsize=11, color='#e74c3c', fontweight='bold')
axes[1].grid(True, alpha=0.3)

# Estacionalidad
axes[2].plot(t, estacionalidad, color='#27ae60', linewidth=2.5)
axes[2].set_ylabel('Estacionalidad', fontsize=10, color='#27ae60')
axes[2].axhline(y=0, color='gray', ls='--', alpha=0.5)
axes[2].text(45, 10, '↻ Se repite', fontsize=11, color='#27ae60', fontweight='bold')
axes[2].grid(True, alpha=0.3)

# Ruido
axes[3].plot(t, ruido, color='#8e44ad', linewidth=1, alpha=0.7)
axes[3].fill_between(t, ruido, 0, alpha=0.2, color='#8e44ad')
axes[3].set_ylabel('Ruido', fontsize=10, color='#8e44ad')
axes[3].axhline(y=0, color='gray', ls='--', alpha=0.5)
axes[3].set_xlabel('Mes', fontsize=12)
axes[3].text(45, 5, '~ Aleatorio', fontsize=11, color='#8e44ad', fontweight='bold')
axes[3].grid(True, alpha=0.3)

# Fórmula
fig.text(0.5, -0.01, 'Dato observado = Tendencia + Estacionalidad + Ruido',
         ha='center', fontsize=13, fontweight='bold',
         bbox=dict(facecolor='lightyellow', edgecolor='gray', boxstyle='round,pad=0.5'))

plt.tight_layout()
plt.show()
```

## Comparación directa: los tres conceptos lado a lado

Ahora que entendemos cada concepto por separado, veámoslos juntos para que quede clara la diferencia.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap

fig, axes = plt.subplots(1, 3, figsize=(20, 5.5))

# ============ PREDICCIÓN ============
ax = axes[0]
np.random.seed(42)
x_pred = np.random.uniform(1, 10, 30)
y_pred = 3 * x_pred + 5 + np.random.randn(30) * 3
ax.scatter(x_pred, y_pred, c='#3498db', s=60, edgecolors='k', alpha=0.7)
coef = np.polyfit(x_pred, y_pred, 1)
x_line = np.linspace(0, 11, 100)
ax.plot(x_line, np.poly1d(coef)(x_line), 'r--', lw=2)
ax.scatter(7, np.poly1d(coef)(7), c='#f39c12', s=300, marker='★', edgecolors='k', zorder=10)
ax.set_title('PREDICCIÓN\n¿Cuánto? → un NÚMERO', fontsize=14, fontweight='bold', color='#2980b9')
ax.set_xlabel('Variable de entrada', fontsize=11)
ax.set_ylabel('Valor a predecir (número)', fontsize=11)
ax.text(2, 33, 'Respuesta:\n"$26,500"\n(valor continuo)', fontsize=10, fontweight='bold',
        bbox=dict(facecolor='#ebf5fb', edgecolor='#2980b9', boxstyle='round,pad=0.5'))
ax.grid(True, alpha=0.3)

# ============ CLASIFICACIÓN ============
ax = axes[1]
np.random.seed(42)
from sklearn.datasets import make_blobs
X_blobs, y_blobs = make_blobs(n_samples=60, centers=3, cluster_std=1.2, random_state=42)

cmap_light = ListedColormap(['#FFCCCC', '#CCFFCC', '#CCCCFF'])
cmap_bold = ListedColormap(['#e74c3c', '#27ae60', '#3498db'])

from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_blobs, y_blobs)
h = 0.1
x_min, x_max = X_blobs[:,0].min()-1, X_blobs[:,0].max()+1
y_min, y_max = X_blobs[:,1].min()-1, X_blobs[:,1].max()+1
xx, yy = np.meshgrid(np.arange(x_min, x_max, h), np.arange(y_min, y_max, h))
Z = knn.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)
ax.contourf(xx, yy, Z, alpha=0.3, cmap=cmap_light)
ax.scatter(X_blobs[:,0], X_blobs[:,1], c=y_blobs, cmap=cmap_bold, edgecolors='k', s=60)
ax.scatter(1, -2, c='lime', s=300, marker='★', edgecolors='k', zorder=10)
ax.set_title('CLASIFICACIÓN\n¿Qué es? → una CATEGORÍA', fontsize=14, fontweight='bold', color='#27ae60')
ax.set_xlabel('Característica 1', fontsize=11)
ax.set_ylabel('Característica 2', fontsize=11)
ax.text(-6, 4, 'Respuesta:\n"Grupo B"\n(etiqueta discreta)', fontsize=10, fontweight='bold',
        bbox=dict(facecolor='#eafaf1', edgecolor='#27ae60', boxstyle='round,pad=0.5'))
ax.grid(True, alpha=0.3)

# ============ PRONÓSTICO ============
ax = axes[2]
np.random.seed(42)
t = np.arange(1, 25)
datos = 50 + t * 0.8 + 10 * np.sin(2*np.pi*t/12) + np.random.randn(24) * 3
t_fut = np.arange(25, 31)
datos_fut = 50 + t_fut * 0.8 + 10 * np.sin(2*np.pi*t_fut/12)

ax.plot(t, datos, 'o-', color='#3498db', markersize=4, lw=1.5, label='Pasado')
ax.plot(t_fut, datos_fut, 's--', color='#e74c3c', markersize=6, lw=2, label='Futuro')
inc = np.arange(1,7) * 3
ax.fill_between(t_fut, datos_fut-inc, datos_fut+inc, color='#e74c3c', alpha=0.15)
ax.axvline(x=24.5, color='gray', ls=':', lw=2)
ax.text(22, 83, 'HOY', fontsize=10, fontweight='bold', color='gray',
        bbox=dict(facecolor='white', edgecolor='gray', boxstyle='round'))
ax.set_title('PRONÓSTICO\n¿Qué pasará? → FUTURO en el tiempo', fontsize=14, fontweight='bold', color='#e74c3c')
ax.set_xlabel('Tiempo (meses)', fontsize=11)
ax.set_ylabel('Valor', fontsize=11)
ax.text(5, 80, 'Respuesta:\n"En marzo: 72"\n(valor futuro)', fontsize=10, fontweight='bold',
        bbox=dict(facecolor='#fdedec', edgecolor='#e74c3c', boxstyle='round,pad=0.5'))
ax.legend(fontsize=9)
ax.grid(True, alpha=0.3)

plt.suptitle('Los Tres Conceptos Lado a Lado', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

fig, ax = plt.subplots(figsize=(12, 7))
ax.axis('off')
ax.set_xlim(-3, 13)
ax.set_ylim(-1, 8)

ax.set_title('Las Zonas Grises: ¿Dónde se cruzan los conceptos?', fontsize=16, fontweight='bold')

# Círculos (Venn simplificado)
from matplotlib.patches import Ellipse

# Predicción (grande, engloba pronóstico)
e_pred = Ellipse((3.5, 4), 7, 5, facecolor='#3498db', alpha=0.15, edgecolor='#2980b9', lw=3)
ax.add_patch(e_pred)
ax.text(1.5, 6.2, 'PREDICCIÓN', fontsize=14, fontweight='bold', color='#2980b9')

# Pronóstico (dentro de predicción)
e_pron = Ellipse((4.5, 3.5), 4, 3, facecolor='#e74c3c', alpha=0.15, edgecolor='#e74c3c', lw=3)
ax.add_patch(e_pron)
ax.text(3.7, 4.6, 'PRONÓSTICO', fontsize=12, fontweight='bold', color='#e74c3c')
ax.text(4.5, 3.5, '(predicción\n+ tiempo)', fontsize=10, ha='center', color='#c0392b',
        fontstyle='italic')

# Clasificación (separada)
e_clas = Ellipse((9.5, 4), 5, 5, facecolor='#27ae60', alpha=0.15, edgecolor='#27ae60', lw=3)
ax.add_patch(e_clas)
ax.text(8.5, 6.2, 'CLASIFICACIÓN', fontsize=14, fontweight='bold', color='#27ae60')

# Ejemplos dentro de cada zona
# Predicción pura
ax.text(1.2, 4, '• Precio de casa\n• Salario esperado\n• Calorías de plato',
        fontsize=8, color='#2c3e50')

# Pronóstico
ax.text(3.5, 2.5, '• Ventas futuras\n• Clima de mañana',
        fontsize=8, color='#2c3e50', ha='center')

# Clasificación pura
ax.text(9.5, 4, '• Spam/No spam\n• Gato/Perro\n• Enfermo/Sano\n• Fraude/Legítimo',
        fontsize=8, color='#2c3e50', ha='center')

# Zona gris: intersección predicción-clasificación
ax.annotate('Zona gris: "¿Aprobará\nel alumno?" es clasificación\n(Sí/No) basada en predicción\n(nota estimada vs umbral)',
            xy=(6.5, 4), xytext=(6, 0.5),
            fontsize=9, ha='center', color='#8e44ad', fontweight='bold',
            arrowprops=dict(arrowstyle='->', color='#8e44ad', lw=1.5),
            bbox=dict(facecolor='#f5eef8', edgecolor='#8e44ad', boxstyle='round,pad=0.3'))

plt.tight_layout()
plt.show()
```

### Casos que pueden confundir

| Caso | ¿Qué es realmente? | ¿Por qué? |
|---|---|---|
| "¿Lloverá mañana? Sí/No" | **Pronóstico** + **Clasificación** | Tiene eje temporal (mañana) y la respuesta es una categoría (sí/no) |
| "¿Cuántos mm lloverá mañana?" | **Pronóstico** (+ predicción) | Tiene eje temporal y la respuesta es un número |
| "¿Este tumor es benigno o maligno?" | **Clasificación** | No hay eje temporal, la respuesta es una categoría |
| "¿Cuánto medirá este niño de adulto?" | **Predicción** | La respuesta es un número, no depende de una serie temporal |
| "¿Cuánto valdrá Bitcoin en diciembre?" | **Pronóstico** | Depende del patrón temporal del precio |

## Ejemplo integrado: el mismo problema visto desde las 3 perspectivas

Imaginemos que trabajamos en un **hospital** y tenemos datos de pacientes. Podemos hacer las 3 tareas con los mismos datos:

```python
import numpy as np
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(20, 5.5))

np.random.seed(42)

# ============ PREDICCIÓN ============
# ¿Cuántos días estará internado este paciente?
ax = axes[0]
edad = np.random.uniform(20, 80, 40)
dias = 1 + (edad - 20) * 0.15 + np.random.randn(40) * 1.5
dias = np.maximum(1, dias)

ax.scatter(edad, dias, c='#3498db', s=60, edgecolors='k', alpha=0.7)
coef = np.polyfit(edad, dias, 1)
x_l = np.linspace(15, 85, 100)
ax.plot(x_l, np.poly1d(coef)(x_l), 'r--', lw=2)

# Paciente nuevo
pac_edad = 55
pac_dias = np.poly1d(coef)(pac_edad)
ax.scatter(pac_edad, pac_dias, c='#f39c12', s=300, marker='★', edgecolors='k', zorder=10)
ax.annotate(f'Paciente de {pac_edad} años\n→ {pac_dias:.1f} días',
            xy=(pac_edad, pac_dias), xytext=(pac_edad-20, pac_dias+3),
            fontsize=10, fontweight='bold', color='#e67e22',
            arrowprops=dict(arrowstyle='->', color='#e67e22', lw=2),
            bbox=dict(facecolor='#fef9e7', edgecolor='#e67e22', boxstyle='round'))

ax.set_xlabel('Edad del paciente', fontsize=11)
ax.set_ylabel('Días de internación', fontsize=11)
ax.set_title('PREDICCIÓN\n¿Cuántos días estará internado?', fontsize=13, fontweight='bold', color='#2980b9')
ax.grid(True, alpha=0.3)

# ============ CLASIFICACIÓN ============
# ¿Tiene riesgo alto o bajo?
ax = axes[1]

n = 25
# Riesgo bajo
presion_baja = np.random.normal(110, 10, n)
colesterol_bajo = np.random.normal(180, 20, n)
# Riesgo alto
presion_alta = np.random.normal(150, 12, n)
colesterol_alto = np.random.normal(260, 25, n)

ax.scatter(presion_baja, colesterol_bajo, c='#27ae60', s=80, edgecolors='k',
           label='Riesgo bajo', marker='o', alpha=0.7)
ax.scatter(presion_alta, colesterol_alto, c='#e74c3c', s=80, edgecolors='k',
           label='Riesgo alto', marker='s', alpha=0.7)

# Frontera
x_f = np.linspace(80, 180, 100)
y_f = -1.5 * x_f + 420
ax.plot(x_f, y_f, 'k--', lw=2, alpha=0.5)
ax.fill_between(x_f, y_f, 350, alpha=0.05, color='red')
ax.fill_between(x_f, y_f, 100, alpha=0.05, color='green')

# Paciente nuevo
pac_p, pac_c = 135, 230
ax.scatter(pac_p, pac_c, c='#f39c12', s=300, marker='★', edgecolors='k', zorder=10)
ax.annotate('¿Riesgo alto\no bajo?',
            xy=(pac_p, pac_c), xytext=(pac_p+10, pac_c-40),
            fontsize=10, fontweight='bold', color='#e67e22',
            arrowprops=dict(arrowstyle='->', color='#e67e22', lw=2),
            bbox=dict(facecolor='#fef9e7', edgecolor='#e67e22', boxstyle='round'))

ax.set_xlabel('Presión arterial (mmHg)', fontsize=11)
ax.set_ylabel('Colesterol (mg/dL)', fontsize=11)
ax.set_title('CLASIFICACIÓN\n¿Tiene riesgo alto o bajo?', fontsize=13, fontweight='bold', color='#27ae60')
ax.legend(fontsize=10)
ax.grid(True, alpha=0.3)

# ============ PRONÓSTICO ============
# ¿Cuántos pacientes llegarán la próxima semana?
ax = axes[2]

dias_sem = np.arange(1, 29)  # 4 semanas
# Patrón semanal: más pacientes lunes, menos el fin de semana
patron_semanal = np.array([50, 45, 42, 40, 35, 20, 15])  # lun-dom
pacientes = np.tile(patron_semanal, 4) + np.random.randn(28) * 5
pacientes = np.maximum(5, pacientes)

dias_fut = np.arange(29, 36)  # próxima semana
pac_fut = patron_semanal + np.array([2, 2, 2, 2, 2, 2, 2])  # leve tendencia al alza

ax.plot(dias_sem, pacientes, 'o-', color='#3498db', markersize=4, lw=1.5, label='Últimas 4 semanas')
ax.plot(dias_fut, pac_fut, 's--', color='#e74c3c', markersize=6, lw=2, label='Próxima semana')
inc = np.arange(1, 8) * 2
ax.fill_between(dias_fut, pac_fut-inc, pac_fut+inc, color='#e74c3c', alpha=0.15)
ax.axvline(x=28.5, color='gray', ls=':', lw=2)
ax.text(28.5, 58, 'HOY', fontsize=9, fontweight='bold', ha='center', color='gray',
        bbox=dict(facecolor='white', edgecolor='gray', boxstyle='round'))

# Marcar patrón semanal
for i in range(4):
    ax.axvspan(i*7+6-0.5, i*7+7+0.5, alpha=0.05, color='gray')  # fines de semana
ax.axvspan(33.5, 35.5, alpha=0.05, color='gray')

ax.set_xlabel('Día', fontsize=11)
ax.set_ylabel('Pacientes en urgencias', fontsize=11)
ax.set_title('PRONÓSTICO\n¿Cuántos pacientes la próxima semana?', fontsize=13, fontweight='bold', color='#e74c3c')
ax.legend(fontsize=9)
ax.grid(True, alpha=0.3)

plt.suptitle('El mismo dominio (Hospital), tres tareas diferentes',
             fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

print("Los mismos datos del hospital permiten hacer las 3 tareas:")
print("  PREDICCIÓN:     ¿Cuántos días estará internado? → 6.3 días")
print("  CLASIFICACIÓN:  ¿Tiene riesgo alto o bajo? → Riesgo alto")
print("  PRONÓSTICO:     ¿Cuántos pacientes vendrán el lunes? → 52 pacientes")
```

## Resumen Final

Para recordar fácilmente la diferencia, piensa en estas **3 preguntas**:

1. **¿La respuesta es un NÚMERO o una CATEGORÍA?**
   - Número → **Predicción** (o pronóstico)
   - Categoría → **Clasificación**

2. **¿Importa el TIEMPO?**
   - Sí, los datos están ordenados en el tiempo → **Pronóstico**
   - No, cada dato es independiente → **Predicción** o **Clasificación**

3. **¿Estamos mirando hacia el FUTURO?**
   - Sí → **Pronóstico**
   - No necesariamente → **Predicción** o **Clasificación**

```python
import matplotlib.pyplot as plt

# Diagrama de flujo para decidir qué tipo de tarea es
fig, ax = plt.subplots(figsize=(14, 8))
ax.axis('off')
ax.set_xlim(0, 14)
ax.set_ylim(0, 9)

# Nodo raíz
rect = plt.Rectangle((4.5, 7.5), 5, 1, facecolor='#ecf0f1', edgecolor='#2c3e50',
                       linewidth=2, zorder=5)
ax.add_patch(rect)
ax.text(7, 8, 'Tengo un problema de ML.\n¿Qué tipo de tarea es?', ha='center', va='center',
        fontsize=11, fontweight='bold')

# Pregunta 1
diamond1 = plt.Polygon([(7, 6.8), (9.5, 5.5), (7, 4.2), (4.5, 5.5)],
                        facecolor='#fef9e7', edgecolor='#f39c12', linewidth=2, zorder=5)
ax.add_patch(diamond1)
ax.text(7, 5.5, '¿La respuesta\nes un número\no una categoría?', ha='center', va='center', fontsize=9)

# Flecha raíz → pregunta 1
ax.annotate('', xy=(7, 6.8), xytext=(7, 7.5),
            arrowprops=dict(arrowstyle='->', color='#2c3e50', lw=2))

# Rama: Categoría → CLASIFICACIÓN
rect_clasif = plt.Rectangle((10.5, 4.8), 3, 1.2, facecolor='#d5f5e3', edgecolor='#27ae60',
                              linewidth=3, zorder=5, joinstyle='round')
ax.add_patch(rect_clasif)
ax.text(12, 5.4, 'CLASIFICACIÓN', ha='center', va='center',
        fontsize=13, fontweight='bold', color='#27ae60')
ax.annotate('', xy=(10.5, 5.5), xytext=(9.5, 5.5),
            arrowprops=dict(arrowstyle='->', color='#27ae60', lw=2))
ax.text(10, 5.9, 'Categoría', fontsize=9, fontweight='bold', color='#27ae60')

# Rama: Número → Pregunta 2
diamond2 = plt.Polygon([(4, 3.8), (6.5, 2.5), (4, 1.2), (1.5, 2.5)],
                        facecolor='#fef9e7', edgecolor='#f39c12', linewidth=2, zorder=5)
ax.add_patch(diamond2)
ax.text(4, 2.5, '¿Los datos tienen\nun orden temporal\ny miramos al futuro?', ha='center', va='center', fontsize=9)

ax.annotate('', xy=(5.5, 4.2), xytext=(5.5, 4.7),
            arrowprops=dict(arrowstyle='->', color='#2980b9', lw=2))
ax.text(4.7, 4.4, 'Número', fontsize=9, fontweight='bold', color='#2980b9')

# Curva de P1 a P2
ax.annotate('', xy=(4, 3.8), xytext=(5.5, 4.2),
            arrowprops=dict(arrowstyle='->', color='#2980b9', lw=2, connectionstyle='arc3,rad=0.3'))

# Rama: Sí → PRONÓSTICO
rect_pron = plt.Rectangle((7, 1.8), 3, 1.2, facecolor='#fdedec', edgecolor='#e74c3c',
                            linewidth=3, zorder=5)
ax.add_patch(rect_pron)
ax.text(8.5, 2.4, 'PRONÓSTICO', ha='center', va='center',
        fontsize=13, fontweight='bold', color='#e74c3c')
ax.annotate('', xy=(7, 2.5), xytext=(6.5, 2.5),
            arrowprops=dict(arrowstyle='->', color='#e74c3c', lw=2))
ax.text(6.7, 2.9, 'Sí', fontsize=10, fontweight='bold', color='#e74c3c')

# Rama: No → PREDICCIÓN
rect_pred = plt.Rectangle((0.3, 0), 3, 1.2, facecolor='#ebf5fb', edgecolor='#2980b9',
                            linewidth=3, zorder=5)
ax.add_patch(rect_pred)
ax.text(1.8, 0.6, 'PREDICCIÓN', ha='center', va='center',
        fontsize=13, fontweight='bold', color='#2980b9')
ax.annotate('', xy=(2.5, 1.2), xytext=(3, 1.6),
            arrowprops=dict(arrowstyle='->', color='#2980b9', lw=2))
ax.text(2, 1.5, 'No', fontsize=10, fontweight='bold', color='#2980b9')

ax.set_title('Diagrama de Flujo: ¿Qué tipo de tarea tengo?', fontsize=16, fontweight='bold', pad=15)

plt.tight_layout()
plt.show()
```
